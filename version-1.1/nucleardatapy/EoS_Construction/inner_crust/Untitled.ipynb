```python
import numpy as np
```

```python


# Neutron mass in MeV (PDG value)
M_N = 939.5654219

# Coefficients from Table 4
COEFFS = {
    "ground_state": np.array([
        -4.0,                    # c0
         2.8822899e-1,           # c1
         5.9150523e-1,           # c2
         9.0185940e-2,           # c3
        -1.1025614e-1,           # c4
         2.9377479e-2,           # c5
        -3.2618465e-3,           # c6
         1.3543555e-4            # c7
    ]),
    "uniform_gas": np.array([
        -4.0,                    # c0
         1.4821424,              # c1
        -4.0373482e-2,           # c2
         6.0455728e-2,           # c3
        -1.5307639e-2,           # c4
         3.4774416e-3,           # c5
        -4.3627154e-4,           # c6
         2.3383473e-5            # c7
    ])
}

def E_T(n_b, configuration="ground_state"):
    """
    Calculate total energy per baryon E_T (MeV) for a given baryon density n_b.
    
    Parameters
    ----------
    n_b : float or array-like
        Baryon density in baryons per cm³.
    configuration : str
        "ground_state" or "uniform_gas".
    
    Returns
    -------
    float or ndarray
        E_T in MeV.
    """
    c = COEFFS[configuration]
    x = np.log(n_b * 1e-35)
    
    # sum_{I=1 to 7} c_I * x**(I-1)
    poly = sum(c[I] * x**(I-1) for I in range(1, 8))
    
    return M_N + c[0] + np.exp(poly) 

def Pres(n_b, configuration="ground_state"):
    """
    Calculate total energy per baryon E_T (MeV) for a given baryon density n_b.
    
    Parameters
    ----------
    n_b : float or array-like
        Baryon density in baryons per cm³.
    configuration : str
        "ground_state" or "uniform_gas".
    
    Returns
    -------
    float or ndarray
        E_T in MeV.
    """
    c = COEFFS[configuration]
    x = np.log(n_b * 1e-35)
    
    # sum_{I=1 to 7} c_I * x**(I-1)
    poly1 = sum(c[I] * x**(I-1) for I in range(1, 8))
    
    
    poly_2 = sum(c[I]*(I-1) * x**(I-2) for I in range(1, 8))
    poly_3 = sum(c[I]*(I-1) *(I-2) * x**(I-3) for I in range(1, 8))

    p1= np.exp(poly1) * poly_2 * n_b
    
    
    
    return p1
```
